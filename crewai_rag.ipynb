```python
import json
from typing import List
import chromadb
from crewai import Agent, Task, Crew, LLM
from crewai.tools import tool
from langchain_huggingface import HuggingFaceEmbeddings
import os

# Set environment variable to avoid OPENAI_API_KEY error
os.environ["OPENAI_API_KEY"] = "API KEY"
# ============================================================================
# READ FILES
# ============================================================================

with open("./sections.txt", "r") as file2:
    act_text = file2.read()

act_text

document = act_text.split("=")
clean_document = []
for i in document:
    if "" is not i:
        clean_document.append(i.replace("\n", " "))
    else:
        continue

with open("./sections_metadata.json", "r") as file:
    metadata = json.load(file)

print(metadata)

act_metadata = []
ids = []
for i, sections in enumerate(metadata["sections"]):
    act_metadata.append({"index": i,
                        "source": "Companies Act 2017",
                        **sections})
    ids.append(str(i))

# ============================================================================
# CONFIGURATION
# ============================================================================

OPENROUTER_API_KEY = "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"

embedder_config = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    encode_kwargs={"normalize_embeddings": True}
)

# ============================================================================
# PERSISTENT RAG CLASS
# ============================================================================

class PersistentRAG:
    
    def __init__(self, sections_file, metadata_file):
        self.chroma_client = chromadb.PersistentClient(path="./chroma_db")
        self.embedder = embedder_config
        self.collection_name = "companies_act_2017"
        
        # Load data
        with open(sections_file, "r") as f:
            act_text = f.read()
        
        document = act_text.split("=")
        self.sections_text = []
        for i in document:
            if i.strip():  # Changed from "" != i to i.strip()
                self.sections_text.append(i.replace("\n", " ").strip())
        
        with open(metadata_file, "r") as f:
            metadata = json.load(f)
        
        # Ensure metadata matches sections length
        sections_data = metadata["sections"][:len(self.sections_text)]  # Trim to match
        
        self.metadata_list = []
        self.ids = []
        for i, sections in enumerate(sections_data):
            self.metadata_list.append({
                "index": i,
                "source": "Companies Act 2017",
                **sections
            })
            self.ids.append(str(i))
        
        # CRITICAL: Ensure all lists have same length
        min_length = min(len(self.sections_text), len(self.metadata_list), len(self.ids))
        self.sections_text = self.sections_text[:min_length]
        self.metadata_list = self.metadata_list[:min_length]
        self.ids = self.ids[:min_length]
        
        print(f"✓ Loaded {len(self.sections_text)} sections")
        print(f"✓ Sections: {len(self.sections_text)}, Metadata: {len(self.metadata_list)}, IDs: {len(self.ids)}")
        
        # Initialize collection
        self.initialize_collection()

    def initialize_collection(self):
        try:
            self.collection = self.chroma_client.get_collection(name=self.collection_name)
            print("✓ Collection existed and loaded successfully")
            
            # Check if already populated
            count = self.collection.count()
            if count == 0:
                print("⚠️  Collection is empty, populating...")
                self.populate_collection()
            else:
                print(f"✓ Collection contains {count} documents")

        except Exception:
            print("⚠️  Creating new collection...")
            self.collection = self.chroma_client.create_collection(
                name=self.collection_name,
                metadata={"description": "Companies Act 2017"}
            )
            self.populate_collection()
        
    def populate_collection(self):
        batch_size = 100

        for i in range(0, len(self.sections_text), batch_size):
            section_batch = self.sections_text[i:i+batch_size]
            meta_batch = self.metadata_list[i:i+batch_size]
            ids_batch = self.ids[i:i+batch_size]
            
            # Generate embeddings
            embeddings = self.embedder.embed_documents(section_batch)
            
            # Validation check
            assert len(section_batch) == len(meta_batch) == len(ids_batch) == len(embeddings), \
                f"Length mismatch: docs={len(section_batch)}, meta={len(meta_batch)}, ids={len(ids_batch)}, emb={len(embeddings)}"
            
            self.collection.add(
                documents=section_batch,
                metadatas=meta_batch,
                ids=ids_batch,
                embeddings=embeddings
            )
            
            print(f"✓ Added batch {i//batch_size + 1}: {len(section_batch)} documents")
        
        print(f"✓ Populated collection with {len(self.sections_text)} documents")
        
    def retrieve_results(self, query, top_k=5) -> List[dict]:
        embed_query = self.embedder.embed_query(query)
        result = self.collection.query(
            query_embeddings=[embed_query],
            n_results=top_k,
            include=["documents", "metadatas", "distances"]
        )
        
        retrieved = []

        if result["documents"] and len(result["documents"][0]) > 0:
            for i in range(len(result["documents"][0])):
                retrieved.append({
                    "text": result["documents"][0][i],
                    "metadata": result["metadatas"][0][i],
                    "score": 1 - result["distances"][0][i]  # Convert distance to similarity
                })
        
        return retrieved

    def reset_db(self):
        try:
            self.chroma_client.delete_collection(name=self.collection_name)
            print("✓ Database deleted successfully")
            self.initialize_collection()
        except Exception as e:
            print(f"⚠️  Error deleting database: {e}")
            self.initialize_collection()

# ============================================================================
# INITIALIZE RAG SYSTEM
# ============================================================================

RESET_DB = False  # Set to True to rebuild embeddings from scratch

if RESET_DB:
    print("⚠️  RESET MODE: Will rebuild entire database\n")
    temp_rag = PersistentRAG('sections.txt', 'sections_metadata.json')
    temp_rag.reset_db()
    rag_system = temp_rag
else:
    # Normal initialization - will load existing DB if available
    rag_system = PersistentRAG('sections.txt', 'sections_metadata.json')

print("\n" + "="*80)
print("RAG SYSTEM READY")
print("="*80 + "\n")

# ============================================================================
# TOOL DEFINITION
# ============================================================================

@tool
def rag_retrieval_tool(query: str) -> str:
    """Retrieve relevant legal sections from the Companies Act 2017 vector database."""
    results = rag_system.retrieve_results(query)

    output = "RETRIEVED FROM VECTOR DATABASE:\n\n"
    for i, result in enumerate(results, 1):
        output += f"--- RESULT {i} (Similarity: {result['score']:.3f}) ---\n"
        output += f"{result['text']}\n"
        if result['metadata']:
            output += f"Metadata: {result['metadata']}\n"
        output += "\n"
    
    return output

# ============================================================================
# LLM CONFIGURATION
# ============================================================================

llm_AO_agent = LLM(
    model="openrouter/tngtech/deepseek-r1t2-chimera:free",
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
    temperature=0.4,
    max_tokens=2000,
)

llm_Reviewer_agent = LLM(
    model="openrouter/tngtech/deepseek-r1t2-chimera:free",
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
    temperature=0.3,
    max_tokens=2000,
)

# ============================================================================
# AGENTS
# ============================================================================

AO_agent = Agent(
    role="Adjudication Officer",
    goal="Analyze legal cases using The Companies Act 2017 via vector database retrieval.",
    backstory="""You are a professional Adjudication Officer.
    
    CRITICAL: Use 'rag_retrieval_tool' which queries a persistent vector database.
    The database contains embeddings of all legal sections for fast semantic search.
    
    Process:
    1. Use rag_retrieval_tool with specific queries
    2. Analyze retrieved sections from database
    3. Cite specific provisions
    """,
    tools=[rag_retrieval_tool],
    llm=llm_AO_agent,
    allow_delegation=True,
    verbose=True,
    memory=True
)

Reviewer_agent = Agent(
    role="Legal Reviewer",
    goal="Verify analysis by cross-checking with vector database.",
    backstory="""You verify legal analyses using database queries.
    Use rag_retrieval_tool to validate all citations.""",
    tools=[rag_retrieval_tool],
    llm=llm_Reviewer_agent,
    allow_delegation=False,
    verbose=True,
    memory=True
)

# ============================================================================
# TASKS
# ============================================================================

legal_task = Task(
    description="""
    Analyze using vector database retrieval:
    
    **Case:** {case}
    
    **Queries to run:**
    1. "annual return filing requirements and penalties"
    2. "books of accounts maintenance Companies Act"
    3. "annual general meeting AGM requirements"
    4. "register of members requirements"
    5. "penalties non-compliance SECP"
    
    Provide analysis with database-backed citations.
    """,
    expected_output="""
    ## Legal Analysis (Database-Backed)
    
    ### Case Summary
    ### Violations (with DB sources)
    ### Analysis
    ### Penalties
    ### Recommendations
    """,
    agent=AO_agent,
    output_file="db_rag_report.md"
)

review_task = Task(
    description="""
    Review by querying database to verify citations.
    Check for missing provisions using database search.
    """,
    expected_output="""
    ## Review Report
    
    **Database Verification**: [Complete]
    **Findings**: [From DB queries]
    **Approval**: [Yes/No]
    """,
    agent=Reviewer_agent,
    output_file="db_rag_reviewed.md",
    context=[legal_task]
)

# ============================================================================
# CREW
# ============================================================================

crew = Crew(
    agents=[AO_agent, Reviewer_agent],
    tasks=[legal_task, review_task],
    verbose=True,
    memory=True,
    process="sequential"
)

# ============================================================================
# EXECUTE
# ============================================================================

if __name__ == "__main__":
    print("="*80)
    print("RUNNING ANALYSIS WITH PERSISTENT VECTOR DATABASE")
    print("="*80)
    print("\n✓ Embeddings stored in: ./chroma_db/")
    print("✓ Database persists across runs")
    print("✓ No re-encoding needed on restart\n")
    print("="*80 + "\n")
    
    try:
        result = crew.kickoff(inputs={
            "case": """ABC Traders (Pvt) Ltd (incorporated in 2018) has:
            
            1. Failed to file annual returns for 2021, 2022, 2023
            2. Not maintained proper books of accounts
            3. Not held AGMs for 3 years
            4. Not updated register of members
            5. Ignored SECP notices
            
            Analyze violations."""
        })
        
        print("\n" + "="*80)
        print("✓ ANALYSIS COMPLETE")
        print("="*80)
        print("\n📁 Vector database location: ./chroma_db/")
        print("📄 Reports: db_rag_report.md, db_rag_reviewed.md")
        
    except Exception as e:
        print(f"\n✗ Error: {str(e)}")
```
